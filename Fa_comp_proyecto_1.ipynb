```python
from vpython import *
import numpy as np
```

```python
# parametros del sistema
g = 9.8 #m/s^2

#cambiar
posx = 0.0
posy =0.0

angulo = 57
r = 0.5 #viscosidad s^-1
V = 50 #m/s^2



teta = angulo * np.pi / 180 # grados a radianes
Vx = V * np.cos(teta)
Vy = V * np.sin(teta)

def Ax(t, velocidadx):
    return -r*velocidadx

def Ay(t):
    return -g


c = canvas(
    title='Simulación de lanzamiento',
    background = color.black,
    align="left"
)

c.width = 500
c.height = 800
c.center = vec(70,50,0)


s = sphere(
    radius= 2.0,
    pos = vec(posx,posy,0),
    color = color.yellow,
    make_trail=True
)

wall = box(
    pos = vec(70,0,0),
    length = 140,
    width = 10,
    hieght = 10
)

# tiempo de colision
tmax = 20
nmax = 2000

t = np.linspace(0, tmax, nmax)
dt = t[1]-t[0]



g1 = graph( width=500, height=0.4*500, align='right', 
            xtitle='time (s)', ytitle='posicion' )
pos_x = gcurve(label = "posx", legend = True, graph=g1, color = color.blue)
pos_y = gcurve(label = "posy", legend=True, graph=g1, color = color.red)

g2 = graph( width=500, height=0.4*500, align='right',
            xtitle='time (s)', ytitle='velocity' )
V_x = gcurve(label = "Vx",legend = True, graph=g2, color = color.blue)
V_y = gcurve(label = "Vy", legend = True, graph=g2, color= color.red)


for ti in t:
    rate(100)

    posx = posx + Vx * dt + Ax(ti, Vx)*(dt**2)/2
    Vx = Vx + Ax(ti, Vx)*dt
    s.pos.x = posx

    
    posy = posy + Vy * dt + Ay(ti)*(dt**2)/2
    Vy = Vy + Ay(ti)*dt
    s.pos.y = posy


    
    pos_x.plot(ti, posx)
    pos_y.plot(ti, posy)

    V_x.plot(ti, Vx)
    V_y.plot(ti, Vy)
    
    if posy <= 0 and ti > 0.1:
        break
    
```
